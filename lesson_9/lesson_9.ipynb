```python
from random import randint
from time import perf_counter_ns
import pandas as pd
```

## Создание массива

```python
MIN_NUMBER = 10
MAX_NUMBER = 250_000_000

# randint включает обе границы, поэтому возможны шаги 3, 4 и 5.
step = randint(3, 5)

# Верхняя граница range не включается, поэтому прибавляем 1.
sorted_numbers = range(MIN_NUMBER, MAX_NUMBER + 1, step)

print(f"Случайный шаг: {step}")
print(f"Количество элементов: {len(sorted_numbers):,}")
print(f"Первое число: {sorted_numbers[0]:,}")
print(f"Последнее число: {sorted_numbers[-1]:,}")
print("Первые 10 элементов:", list(sorted_numbers[:10]))
```

## Случайные числа

```python
random_numbers = [randint(MIN_NUMBER, MAX_NUMBER) for _ in range(10)]
print("Числа для поиска:")
print(random_numbers)
```

## Функции поиска

```python
def linear_search(array, target):
    """Последовательно ищет target и возвращает его индекс или -1."""
    for index, value in enumerate(array):
        if value == target:
            return index

        # Массив отсортирован: после target искать дальше бессмысленно.
        if value > target:
            return -1

    return -1


def binary_search(array, target):
    """Делит область поиска пополам и возвращает индекс или -1."""
    left = 0
    right = len(array) - 1

    while left <= right:
        middle = (left + right) // 2
        middle_value = array[middle]

        if middle_value == target:
            return middle
        elif middle_value < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1
```

## Замер времени

```python
results = []

for target in random_numbers:
    linear_start = perf_counter_ns()
    linear_index = linear_search(sorted_numbers, target)
    linear_time_ns = perf_counter_ns() - linear_start

    binary_start = perf_counter_ns()
    binary_index = binary_search(sorted_numbers, target)
    binary_time_ns = perf_counter_ns() - binary_start

    # Оба алгоритма должны дать одинаковый результат.
    assert linear_index == binary_index

    results.append({
        "Искомое число": target,
        "Найдено": linear_index != -1,
        "Индекс": linear_index if linear_index != -1 else "—",
        "Линейный поиск, мс": linear_time_ns / 1_000_000,
        "Бинарный поиск, мс": binary_time_ns / 1_000_000,
    })

results_df = pd.DataFrame(results)
results_df.style.format({
    "Искомое число": "{:,}",
    "Линейный поиск, мс": "{:.6f}",
    "Бинарный поиск, мс": "{:.6f}",
})
```

## Итоговые показатели

```python
linear_total = results_df["Линейный поиск, мс"].sum()
binary_total = results_df["Бинарный поиск, мс"].sum()
found_count = int(results_df["Найдено"].sum())

print(f"Найдено чисел: {found_count} из {len(random_numbers)}")
print(f"Общее время линейного поиска: {linear_total:.6f} мс")
print(f"Общее время бинарного поиска: {binary_total:.6f} мс")

if binary_total > 0:
    print(f"Отношение времени: {linear_total / binary_total:,.2f}")
```
