```python
from datetime import datetime, date


class LegalObligation:
    def __init__(self, obligation_number, creditor, debtor, subject, due_date, amount=0):
        self.obligation_number = obligation_number
        self.creditor = creditor
        self.debtor = debtor
        self.subject = subject
        self.amount = amount
        self.due_date = self._parse_date(due_date)
        self.status = "Действует"
        self.evidence = []
        self.performance_history = []

    @staticmethod
    def _parse_date(value):
        """Преобразует строку формата ДД.ММ.ГГГГ в дату."""
        if isinstance(value, date):
            return value
        return datetime.strptime(value, "%d.%m.%Y").date()

    def add_evidence(self, evidence):
        """Добавляет документ или другое доказательство."""
        if evidence not in self.evidence:
            self.evidence.append(evidence)

    def extend_due_date(self, new_due_date):
        """Переносит срок, если новая дата позднее действующего срока."""
        new_due_date = self._parse_date(new_due_date)
        if new_due_date <= self.due_date:
            raise ValueError("Новый срок должен быть позднее текущего")

        old_due_date = self.due_date
        self.due_date = new_due_date

        if self.status == "Просрочено":
            self.status = "Действует"

        print(
            f"Срок изменён с {old_due_date:%d.%m.%Y} "
            f"на {self.due_date:%d.%m.%Y}"
        )

    def check_overdue(self, current_date=None):
        """Возвращает True, если неисполненное обязательство просрочено."""
        if current_date is None:
            current_date = date.today()
        else:
            current_date = self._parse_date(current_date)

        if self.status == "Исполнено":
            return False

        if current_date > self.due_date:
            self.status = "Просрочено"
            return True

        self.status = "Действует"
        return False

    def fulfill(self, description, performance_date=None):
        """Фиксирует исполнение и завершает обязательство."""
        if self.status == "Исполнено":
            print("Обязательство уже исполнено")
            return

        if performance_date is None:
            performance_date = date.today()
        else:
            performance_date = self._parse_date(performance_date)

        self.performance_history.append({
            "date": performance_date,
            "description": description
        })
        self.status = "Исполнено"
        print("Исполнение обязательства зафиксировано")

    def get_info(self):
        """Возвращает основные сведения об обязательстве."""
        return (
            f"Обязательство № {self.obligation_number}\n"
            f"Кредитор: {self.creditor}\n"
            f"Должник: {self.debtor}\n"
            f"Предмет: {self.subject}\n"
            f"Сумма: {self.amount:,.2f} руб.\n"
            f"Срок: {self.due_date:%d.%m.%Y}\n"
            f"Статус: {self.status}"
        )
```

```python
obligation = LegalObligation(
    obligation_number="ОБ-15/2026",
    creditor="ООО «Право»",
    debtor="ООО «Вектор»",
    subject="Оплата юридических услуг",
    due_date="30.09.2026",
    amount=150000
)

obligation.add_evidence("Договор № 15 от 01.09.2026")
obligation.add_evidence("Акт оказанных услуг")

print(obligation.get_info())
print("Просрочено на 01.10.2026:", obligation.check_overdue("01.10.2026"))
print("Текущий статус:", obligation.status)

obligation.extend_due_date("15.10.2026")
print("Просрочено на 10.10.2026:", obligation.check_overdue("10.10.2026"))

obligation.fulfill("Денежные средства перечислены кредитору", "12.10.2026")

print()
print(obligation.get_info())
print("Доказательства:", obligation.evidence)
print("История исполнения:", obligation.performance_history)
```
