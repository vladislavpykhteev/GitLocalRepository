```python
from random import randint
from time import perf_counter

import numpy as np
from numba import njit
```

## Первый массив

```python
ARRAY_LENGTH = 100_000
MIN_VALUE = 1
MAX_VALUE = 1_000_000

# Сначала значения создаются с помощью randint и list comprehension,
# затем преобразуются в числовой массив для быстрой компиляции циклов.
integer_array = np.array(
    [randint(MIN_VALUE, MAX_VALUE) for _ in range(ARRAY_LENGTH)],
    dtype=np.int64,
)

print(f"Количество элементов: {len(integer_array):,}")
print("Первые 10 элементов:", integer_array[:10])
```

## Пузырьковая сортировка

```python
@njit
def bubble_sort(array):
    """Сортирует числовой массив по возрастанию методом пузырька."""
    right_border = len(array)

    while right_border > 1:
        last_swap_index = 0

        for index in range(1, right_border):
            if array[index - 1] > array[index]:
                array[index - 1], array[index] = (
                    array[index],
                    array[index - 1],
                )
                last_swap_index = index

        # Если обменов не было, массив уже отсортирован.
        right_border = last_swap_index

    return array
```

## Сортировка чисел

```python
# Предварительная компиляция функции на маленьком массиве.
bubble_sort(np.array([2, 1], dtype=np.int64))

bubble_start = perf_counter()
bubble_sort(integer_array)
bubble_time = perf_counter() - bubble_start

# Проверяем длину и порядок элементов.
assert len(integer_array) == ARRAY_LENGTH
assert np.all(integer_array[:-1] <= integer_array[1:])

print("Первые 10 элементов после сортировки:", integer_array[:10])
print("Последние 10 элементов после сортировки:", integer_array[-10:])
print(f"Время пузырьковой сортировки: {bubble_time:.3f} сек.")
```

## Массив словарей

```python
dictionary_array = [
    {
        "num_1": randint(MIN_VALUE, MAX_VALUE),
        "num_2": randint(MIN_VALUE, MAX_VALUE),
    }
    for _ in range(ARRAY_LENGTH)
]

assert len(dictionary_array) == ARRAY_LENGTH

print(f"Количество словарей: {len(dictionary_array):,}")
print("Первые 5 словарей до сортировки:")
for item in dictionary_array[:5]:
    print(item)
```

## Сортировка словарей

```python
dictionary_start = perf_counter()
dictionary_array.sort(key=lambda item: (item["num_1"], item["num_2"]))
dictionary_time = perf_counter() - dictionary_start

# Проверяем сортировку каждой соседней пары.
assert all(
    (dictionary_array[index]["num_1"], dictionary_array[index]["num_2"])
    <=
    (dictionary_array[index + 1]["num_1"], dictionary_array[index + 1]["num_2"])
    for index in range(len(dictionary_array) - 1)
)

print("Первые 10 словарей после сортировки:")
for item in dictionary_array[:10]:
    print(item)

print(f"Время встроенной сортировки: {dictionary_time:.6f} сек.")
```

## Результат

```python
print("Проверка завершена успешно.")
print(f"Длина первого массива: {len(integer_array):,}")
print(f"Длина второго массива: {len(dictionary_array):,}")
print(f"Пузырьковая сортировка: {bubble_time:.3f} сек.")
print(f"Встроенная сортировка словарей: {dictionary_time:.6f} сек.")
```
